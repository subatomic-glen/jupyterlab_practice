```python
def compute_mean(values):
    """
    Computes the mean of values in a list and prints the result.
    
    Parameters:
    values (list): A list of numeric values
    
    Returns:
    float: The mean of the values
    """
    mean = sum(values) / len(values)
    print(f"Mean: {mean}")
    return mean

compute_mean([2, 4, 6, 8])
```

$$mean = \frac{\sum(values)}{len(values)}$$
